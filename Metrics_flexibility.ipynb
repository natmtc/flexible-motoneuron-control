```julia

# Read the .mat files and extract the 'dispersion' vectors
file_path1 = "/Users/nmt121/Desktop/DR_87TRIANG_gamma0.7.jld2.mat"
data1 = matread(file_path1)
disp = data1["dispersion"]

file_path2 = "/Users/nmt121/Desktop/DR_87TRIANG_gamma0.jld2.mat"
data2 = matread(file_path2)
dispgamma0 = data2["dispersion"]

# Define x values for plotting
x1 = 2*ones(length(disp))  # x = 1 for all values in disp
x5 = 6 * ones(length(dispgamma0))  # x = 5 for all values in dispgamma0


p = Plots.plot(legend=false, grid=false,ylim=(-2, 25))

# Create the plot
Plots.plot!(x1, dispgamma0[1,:], label="disp @ x=1", seriestype=:scatter, color=:lightsteelblue1,markerstrokewidth=0.5)
Plots.plot!(x5, disp[1,:], label="dispgamma0 @ x=5", seriestype=:scatter, color=:slateblue,markerstrokewidth=0.5)

# Add lines connecting the corresponding values
for i in 1:min(length(disp), length(dispgamma0))
    Plots.plot!([x1[i], x5[i]], [ dispgamma0[i],disp[i]], color=RGBA(200, 0, 150, 0.2), linestyle=:solid, label=false)
end

display(p)
```
